### 1.1 Вероятностное пространство, формула Байеса

### №1 Определить (с обоснованием), зависимы или независимы следующие события:

**(a) Несовместные события.**

Пусть события $A$ и $B$ несовместны, т.е. $A \cap B = \varnothing$.
Тогда
$$
P(A \cap B) = 0.
$$
Если бы $A$ и $B$ были независимы, то должно выполняться
$$
P(A \cap B) = P(A)P(B).
$$
Следовательно,
$$
0 = P(A)P(B),
$$
что возможно только при $P(A)=0$ или $P(B)=0$.

**Вывод:** в общем случае несовместные события зависимы; независимость возможна
лишь в вырожденном случае, когда вероятность одного из событий равна нулю.

---

**(b) События, образующие $\sigma$-алгебру $\Sigma$ в пространстве $(\Omega,\Sigma,P)$.**

Сам факт того, что события принадлежат одной $\sigma$-алгебре,
не определяет их независимость: возможны как независимые,
так и зависимые события.

*Пример зависимых:* возьмём $A \in \Sigma$ такое, что $0 < P(A) < 1$.
Тогда $\overline{A} \in \Sigma$ и
$$
A \cap \overline{A} = \varnothing, \qquad
P(A \cap \overline{A}) = 0, \qquad
P(A)P(\overline{A}) > 0,
$$
то есть события $A$ и $\overline{A}$ зависимы.

*Пример независимых:* пусть $\Omega$ — исходы двух честных подбрасываний монеты,
$$
A = \{\text{орёл на 1-й монете}\}, \quad
C = \{\text{орёл на 2-й монете}\}.
$$
Тогда
$$
P(A)=\frac12, \quad P(C)=\frac12, \quad
P(A \cap C)=\frac14 = P(A)P(C),
$$
то есть $A$ и $C$ независимы.

**Вывод:** принадлежность $\sigma$-алгебре не влечёт
ни независимости, ни зависимости.

---

**(c) События, имеющие одинаковую вероятность.**

Равенство вероятностей $P(A)=P(B)$ само по себе
не гарантирует независимость.

Контрпример: пусть $A=B$ и $0<P(A)<1$ (например, $P(A)=\tfrac12$).
Тогда
$$
P(A \cap B) = P(A) \neq P(A)P(B) = P(A)^2,
$$
следовательно, события $A$ и $B$ зависимы.

**Вывод:** события с одинаковой вероятностью в общем случае
могут быть зависимыми
(и могут быть независимыми — но это не следует из равенства вероятностей).



### №2 Опыт, состоящий в независимом подбрасывании двух симметричных монет.
Вводятся следующие события:

- \(A\) — выпадение герба на первой монете;
- \(B\) — выпадение решки на первой монете;
- \(C\) — выпадение герба на второй монете;
- \(D\) — выпадение решки на второй монете;
- \(E\) — появление хотя бы одного герба;
- \(F\) — появление хотя бы одной решки;
- \(G\) — появление одного герба и одной решки;
- \(H\) — отсутствие гербов;
- \(K\) — появление двух гербов.

Требуется определить, каким событиям из данного списка равносильны следующие выражения:
- (a) \(A + C\)
- (b) \(AC\)
- (c) \(EF\)
- (d) \(G + E\)
- (e) \(GE\)
- (f) \(BD\)
- (g) \(E + K\)

---

### Решение

Рассматривается эксперимент независимого подбрасывания двух симметричных монет.

Пространство элементарных исходов:
$$
\Omega = \{ГГ, ГР, РГ, РР\}.
$$
Здесь первая буква обозначает результат первой монеты,
а вторая — результат второй монеты.

События из условия задачи задаются следующими множествами исходов:

$$
\begin{aligned}
A &= \{ГГ, ГР\},\\
B &= \{РГ, РР\},\\
C &= \{ГГ, РГ\},\\
D &= \{ГР, РР\},\\
E &= \{ГГ, ГР, РГ\},\\
F &= \{ГР, РГ, РР\},\\
G &= \{ГР, РГ\},\\
H &= \{РР\},\\
K &= \{ГГ\}.
\end{aligned}
$$

Определим, каким событиям равносильны выражения из условия.

**(a)**  
$$
A + C = \{ГГ, ГР, РГ\} = E
$$
— появление хотя бы одного герба.

**(b)**  
$$
AC = \{ГГ\} = K
$$
— появление двух гербов.

**(c)**  
$$
EF = \{ГР, РГ\} = G
$$
— появление одного герба и одной решки.

**(d)**  
$$
G + E = \{ГГ, ГР, РГ\} = E
$$
— появление хотя бы одного герба.

**(e)**  
$$
GE = \{ГР, РГ\} = G
$$
— появление одного герба и одной решки.

**(f)**  
$$
BD = \{РР\} = H
$$
— непоявление ни одного герба.

**(g)**  
$$
E + K = \{ГГ, ГР, РГ\} = E
$$
— появление хотя бы одного герба.



### №3

Производится выстрел по вращающейся круговой мишени, в которой закрашены два непересекающихся сектора с углом $20^\circ$ каждый. 
Найти вероятность попадания в закрашенную область.

---

### Решение

Пусть \(A\) — событие «попадание в закрашенную область». Вероятность равна отношению площади закрашенной части к площади всей мишени.

Площадь одного сектора с углом $20^\circ$:
$$
S_{\text{сектор}}=\frac{20^\circ}{360^\circ}\,\pi r^2.
$$

Так как закрашено два таких сектора, то
$$
S_{\text{закр}} = 2S_{\text{сектор}} = 2\cdot \frac{20}{360}\,\pi r^2.
$$

Тогда
$$
P(A)=\frac{S_{\text{закр}}}{S_{\text{круга}}}
=\frac{2\cdot \frac{20}{360}\,\pi r^2}{\pi r^2}
=\frac{40}{360}
=\frac{1}{9}.
$$

**Ответ:** $P(A)=\dfrac{1}{9}$.


###  №4

Два парохода независимо друг от друга и равновозможно в течение суток
подходят к одному и тому же причалу.
Время стоянки первого парохода — 1 час, второго — 2 часа.
Найти вероятность того, что одному из них придётся ожидать освобождения причала.

---

### Решение

Пусть \(X\) и \(Y\) — моменты прихода первого и второго пароходов
(в часах от начала суток).
Считаем, что \(X\) и \(Y\) независимы и равномерно распределены на отрезке [0,24].

Первый пароход занимает причал в интервале [X, X+1],  
второй — в интервале [Y, Y+2].

Ожидание возникает тогда, когда эти интервалы пересекаются.

Ожидания не будет, если один из пароходов успевает уйти до прихода другого:
- первый уходит раньше прихода второго;
- второй уходит раньше прихода первого.

Рассмотрим квадрат возможных значений (X,Y) со стороной 24.
Область без ожидания состоит из двух треугольников.

Площадь первого треугольника:
$$
S_1 = \frac{23^2}{2}.
$$

Площадь второго треугольника:
$$
S_2 = \frac{22^2}{2}.
$$

Площадь области без ожидания:
$$
S_{\text{без}} = \frac{23^2 + 22^2}{2}.
$$

Площадь всего квадрата:
$$
S_{\text{всего}} = 24^2.
$$

Вероятность того, что одному из пароходов придётся ожидать освобождения причала:
$$
P = 1 - \frac{S_{\text{без}}}{S_{\text{всего}}}
  = 1 - \frac{23^2 + 22^2}{2 \cdot 24^2}
  = \frac{139}{1152}.
$$

**Ответ:** $P = \frac{139}{1152} \approx 0.121$.

### №5

Самолёт, по которому ведётся стрельба, состоит из трёх различных по уязвимости частей:

(a) кабина лётчика и двигатель;  
(b) топливные баки;  
(c) планер.

Для поражения самолёта достаточно либо одного попадания в первую часть,
либо двух попаданий во вторую часть, либо трёх попаданий в третью часть.

При попадании одного снаряда в самолёт он с вероятностью p1 попадает в первую часть,
с вероятностью p2 — во вторую часть и с вероятностью p3 — в третью часть.
Попадания снарядов по частям самолёта считаются независимыми друг от друга.

Известно, что в самолёт попало m снарядов.
Требуется найти условную вероятность P(A | m) события A — «самолёт поражён»
при m = 1, 2, 3, 4.

---

#### Случай m = 1

При одном попадании самолёт будет поражён только в том случае,
если снаряд попал в первую часть.

Следовательно,
P(A | 1) = p1.

---

#### Случай m = 2

Самолёт не поражён, если:
- оба снаряда попали в третью часть;
- один снаряд попал во вторую часть, а другой — в третью.

Вероятность первого случая равна p3².  
Вероятность второго случая равна 2·p2·p3 (учитываются два возможных порядка попаданий).

Итак, вероятность того, что самолёт не поражён при двух попаданиях, равна:
p3² + 2·p2·p3.

Следовательно,
P(A | 2) = 1 − p3² - 2·p2·p3.

---

#### Случай m = 3

Чтобы самолёт не был поражён:
- в первую часть попаданий быть не должно;
- во вторую часть должно быть ровно одно попадание;
- в третью часть — ровно два попадания.

Другие варианты либо приводят к поражению,
либо невозможны при трёх попаданиях.

Число возможных перестановок таких попаданий равно 3.

Вероятность того, что самолёт не поражён, равна:
3·p2·p3².

Следовательно,
P(A | 3) = 1 − 3·p2·p3².

---

#### Случай m = 4

Если в самолёт попало четыре снаряда, то при отсутствии попаданий в первую часть
и не более одного попадания во вторую часть общее число попаданий не может
превысить трёх.

Следовательно, при четырёх попаданиях самолёт поражается всегда.

Таким образом,
P(A | 4) = 1.

---

### Ответ

P(A | 1) = p1  
P(A | 2) = 1 − p3² - 2·p2·p3  
P(A | 3) = 1 − 3·p2·p3²  
P(A | 4) = 1

### 1.2 Случайный вектор и числовые характеристики

### №1

Пусть
$$
f_\xi(x,y)=\frac{e^{-2|y|}}{\pi(1+x^2)}.
$$

Является ли данная функция плотностью распределения случайного вектора?

---

#### Решение

1. Проверим неотрицательность функции.

Так как
$$
e^{-2|y|} > 0,\quad y \in \mathbb{R},
$$
а также
$$
\pi > 0 \quad \text{и} \quad \frac{1}{1+x^2} > 0,\quad x \in \mathbb{R},
$$
то
$$
f_\xi(x,y) > 0,\quad (x,y) \in \mathbb{R}^2.
$$

2. Проверим условие нормировки.

Рассмотрим интеграл функции по всей плоскости:
$$
\int_{-\infty}^{\infty}\int_{-\infty}^{\infty}
\frac{e^{-2|y|}}{\pi(1+x^2)}\,dx\,dy.
$$

Так как функция представляется в виде произведения множителей,
зависящих отдельно от \(x\) и от \(y\), получаем
$$
\int_{-\infty}^{\infty}\frac{dx}{\pi(1+x^2)}
\cdot
\int_{-\infty}^{\infty} e^{-2|y|}\,dy.
$$

Первый интеграл равен
$$
\int_{-\infty}^{\infty}\frac{dx}{\pi(1+x^2)}
=
\frac{1}{\pi}\left(\arctan x\right)\Big|_{-\infty}^{\infty}
=
\frac{1}{\pi}\left(\frac{\pi}{2}+\frac{\pi}{2}\right)
=1.
$$

Второй интеграл равен
$$
\int_{-\infty}^{\infty} e^{-2|y|}\,dy
=
2\int_{0}^{\infty} e^{-2y}\,dy
=
1.
$$

Следовательно,
$$
\int_{-\infty}^{\infty}\int_{-\infty}^{\infty}
f_\xi(x,y)\,dx\,dy = 1.
$$

---

#### Вывод

Функция $f_{\xi}(x,y)$ неотрицательна на всей плоскости
и её интеграл по всей области определения равен единице.
Следовательно, данная функция является плотностью распределения
случайного вектора.


### №2

Совместное распределение случайных величин ξ и η задано таблицей вероятностей.

Случайная величина ξ принимает значения −1 и 1.  
Случайная величина η принимает значения −1, 0 и 1.

Вероятности заданы следующим образом:

При ξ = −1:
- вероятность того, что η = −1, равна 1/8;
- вероятность того, что η = 0, равна 1/12;
- вероятность того, что η = 1, равна 7/24.

При ξ = 1:
- вероятность того, что η = −1, равна 1/3;
- вероятность того, что η = 0, равна 1/6;
- вероятность того, что η = 1, равна 0.

Требуется:
(a) найти маргинальные распределения ξ и η;  
(b) вычислить математическое ожидание, ковариационную и корреляционную матрицы вектора (ξ, η);  
(c) исследовать ξ и η на независимость и некоррелированность.

---

### Решение

#### (a) Маргинальные распределения

Маргинальное распределение ξ находится суммированием вероятностей по всем значениям η.

Для ξ = −1:
1/8 + 1/12 + 7/24 = 1/2.

Для ξ = 1:
1/3 + 1/6 + 0 = 1/2.

Следовательно, случайная величина ξ принимает значения −1 и 1 с одинаковой вероятностью 1/2.

Маргинальное распределение η находится суммированием вероятностей по всем значениям ξ.

Для η = −1:
1/8 + 1/3 = 11/24.

Для η = 0:
1/12 + 1/6 = 1/4.

Для η = 1:
7/24 + 0 = 7/24.

---
**(b) Найдём соответствующие математические ожидания.**

Математическое ожидание случайной величины η равно −1/6.  
Математическое ожидание случайной величины ξ равно 0.

Для составления ковариационной матрицы вычислим ковариацию и соответствующие дисперсии Dξ и Dη.

Ковариация ξ и η вычисляется как разность между математическим ожиданием произведения ξη и произведением математических ожиданий ξ и η.

Дисперсия ξ вычисляется как разность между математическим ожиданием ξ в квадрате и квадратом математического ожидания ξ.  
Дисперсия η вычисляется как разность между математическим ожиданием η в квадрате и квадратом математического ожидания η.

Составим распределения для случайных величин ξ², η² и ξη.

Распределение ξ²:  
ξ² = 1 с вероятностью 1.

Распределение η²:  

| значение η² | вероятность |
|------------|-------------|
| 0          | 1/4         |
| 1          | 3/4         |

Распределение произведения ξη:  

| значение ξη | вероятность |
|------------|-------------|
| −1         | 15/24       |
| 0          | 1/4         |
| 1          | 1/8         |

Тогда математическое ожидание произведения ξη равно −15/24 + 1/8, что даёт −0.5.

Следовательно:
- дисперсия ξ равна 1;
- дисперсия η равна 13/18;
- ковариация ξ и η равна −0.5.

Составим ковариационную матрицу.

Дисперсия случайной величины ξ равна 1.  
Дисперсия случайной величины η равна 13/18.  
Ковариация ξ и η равна −0.5.

Следовательно, ковариационная матрица случайного вектора (ξ, η) имеет следующий вид:

[  1      −0.5  ]  
[ −0.5   13/18 ]

Для составления корреляционной матрицы найдём коэффициент корреляции между ξ и η.

Коэффициент корреляции равен ковариации, делённой на произведение стандартных отклонений.
В данном случае коэффициент корреляции численно равен приблизительно −0.588.

Теперь составим корреляционную матрицу.

Корреляционная матрица случайного вектора (ξ, η) имеет вид:

[  1     −0.588 ]  
[ −0.588   1    ]


---

**(c) Исследуем ξ и η на некоррелированность и независимость.**

Так как коэффициент корреляции не равен нулю, случайные величины ξ и η являются коррелированными.

Проверим независимость.

Рассмотрим значения ξ = −1 и η = −1.  
Произведение маргинальных вероятностей: 11/24*1/2=11/48.

Совместная вероятность этих значений по таблице равна 1/8.

Так как $1/8 \neq 11/48$, случайные величины ξ и η являются зависимыми.


### №3

Пусть имеются два одинаковых тетраэдра с числами 1, 2, 3, 4 на гранях. Подкидываем оба и смотрим на выпавшие числа ξ1 и ξ2. Зададим следующие случайные величины: φ1 = ξ1+ξ2; φ2 = 1, если $ξ1 \neq ξ2$, и = 0 иначе.

Требуется:
(a) составить таблицу совместного распределения φ1 и φ2;  
(b) найти маргинальные распределения φ1 и φ2;  
(c) вычислить математическое ожидание, ковариационную и корреляционную матрицы вектора (φ1, φ2);  
(d) исследовать φ1 и φ2 на независимость и некоррелированность.

---

### Решение

#### (a) Таблица совместного распределения φ1 и φ2

Ниже приведены вероятности для всех значений φ1 от 2 до 8 и φ2, равного 0 или 1.

Совместное распределение (строки — значения φ2, столбцы — значения φ1):

| φ2 \ φ1 | 2    | 3    | 4    | 5    | 6    | 7    | 8    |
|:------:|:----:|:----:|:----:|:----:|:----:|:----:|:----:|
|   0    | 0    | 0    | 0    | 1/8  | 0    | 1/8  | 0    |
|   1    | 1/16 | 1/8  | 3/16 | 1/8  | 3/16 | 0    | 1/16 |

#### (b) Маргинальные распределения φ1 и φ2

Маргинальное распределение φ1:

| φ1 | 2    | 3    | 4    | 5    | 6    | 7    | 8    |
|---:|:----:|:----:|:----:|:----:|:----:|:----:|:----:|
| P  | 1/16 | 1/8  | 3/16 | 1/4  | 3/16 | 1/8  | 1/16 |

Маргинальное распределение φ2:

| φ2 | 0   | 1   |
|---:|:---:|:---:|
| P  | 1/4 | 3/4 |

#### (c) Математические ожидания, ковариационная и корреляционная матрицы

Математическое ожидание φ1 вычисляется как сумма значений φ1, умноженных на их вероятности. В результате получается значение 5.

Математическое ожидание φ2 равно 3/4, так как значение 1 принимается с вероятностью 3/4.

Для ковариации используем стандартную формулу: ковариация равна математическому ожиданию произведения φ1·φ2 минус произведение математических ожиданий φ1 и φ2.

Для дисперсий используем стандартную формулу: дисперсия равна математическому ожиданию квадрата минус квадрат математического ожидания.

Распределение φ1²:

| φ1² | 4    | 9    | 16   | 25   | 36   | 49   | 64   |
|----:|:----:|:----:|:----:|:----:|:----:|:----:|:----:|
|  P  | 1/16 | 1/8  | 3/16 | 1/4  | 3/16 | 1/8  | 1/16 |

Распределение φ2² (оно совпадает с распределением φ2, так как 0² = 0, 1² = 1):

| φ2² | 0   | 1   |
|----:|:---:|:---:|
|  P  | 1/4 | 3/4 |

Распределение произведения φ1·φ2:

| φ1·φ2 | 0   | 2   | 3   | 4   | 5   | 6   | 7   | 8   |
|------:|:---:|:---:|:---:|:---:|:---:|:---:|:---:|:---:|
|   P   | 1/4 | 0   | 1/8 | 1/8 | 1/4 | 1/8 | 1/8 | 0   |

Из этого получаем следующие значения:
- ковариация φ1 и φ2 равна 0;
- дисперсия φ1 равна 2.5;
- дисперсия φ2 равна 3/16.

Ковариационная матрица вектора (φ1, φ2):

|        | φ1  | φ2   |
|--------|-----:|-----:|
| **φ1** | 2.5 | 0    |
| **φ2** | 0   | 3/16 |

Коэффициент корреляции равен 0, так как ковариация равна 0.

Корреляционная матрица:

|        | φ1 | φ2 |
|--------|---:|---:|
| **φ1** | 1  | 0  |
| **φ2** | 0  | 1  |

#### (d) Независимость и некоррелированность

Так как коэффициент корреляции равен 0, величины φ1 и φ2 являются некоррелированными.

Проверим независимость на конкретном примере. Вероятность события «φ1 = 2 и φ2 = 0» равна 1/16. При независимости она должна была бы равняться произведению вероятностей «φ1 = 2» и «φ2 = 0», то есть (1/16) умножить на (1/4), что равно 1/64. Так как $1/16 \neq 1/64$, величины φ1 и φ2 зависимы.


### №4

Пусть ξ равномерно распределена на отрезке [−π, π], а также заданы случайные величины η1 = cos ξ и η2 = sin ξ.  
Требуется: (a) вычислить математические ожидания, ковариационную и корреляционную матрицы вектора (η1, η2)
(b) исследовать η1 и η2 на независимость и некоррелированность.

---

### Решение

(a) Так как распределение ξ равномерно на симметричном промежутке, средние значения cos ξ и sin ξ равны нулю, поэтому E(η1) = 0 и E(η2) = 0. Среднее значение cos² ξ на полном периоде равно 1/2, и среднее значение sin² ξ также равно 1/2, следовательно E(η1²) = 1/2 и E(η2²) = 1/2. Тогда дисперсии равны: D(η1) = E(η1²) − (E(η1))² = 1/2 и D(η2) = E(η2²) − (E(η2))² = 1/2. Ковариация равна cov(η1, η2) = E(η1η2) − E(η1)E(η2). Так как E(η1) = E(η2) = 0, достаточно найти E(η1η2) = E(cos ξ · sin ξ). Произведение cos ξ · sin ξ является нечётной функцией на симметричном промежутке, поэтому его среднее значение равно 0, значит cov(η1, η2) = 0. Ковариационная матрица вектора (η1, η2) имеет вид: на диагонали стоят значения 1/2 и 1/2, а вне диагонали нули, то есть матрица равна [[1/2, 0], [0, 1/2]]. Коэффициент корреляции равен 0, так как ковариация равна 0, поэтому корреляционная матрица равна единичной: [[1, 0], [0, 1]].

(b) Так как коэффициент корреляции равен 0, величины η1 и η2 некоррелированны. Однако независимыми они не являются, поскольку между ними существует функциональная связь: для любых значений ξ выполняется равенство cos² ξ + sin² ξ = 1. Это означает, что возможные значения пары (η1, η2) всегда лежат на единичной окружности, и потому знание одной величины накладывает ограничение на другую, следовательно независимости нет.


### №5

Найти плотность распределения суммы двух независимых случайных величин ξ и η, если ξ имеет экспоненциальное распределение с параметром 2, а η равномерно распределена на отрезке [0, 1].

---

### Решение

Так как случайные величины независимы, плотность суммы Z = ξ + η находится через свёртку.

Плотность ξ:
- равна 0 при x < 0;
- равна 2·e^(−2x) при x ≥ 0.

Плотность η:
- равна 0 вне отрезка [0, 1];
- равна 1 на отрезке [0, 1].

Плотность суммы Z в точке x равна интегралу по t от 0 до бесконечности от 2·e^(−2t) умножить на плотность η в точке (x − t). Так как η отлична от нуля только на [0, 1], требуется, чтобы одновременно выполнялись условия:
- t ≥ 0;
- x − t принадлежит отрезку [0, 1], то есть 0 ≤ x − t ≤ 1.

Отсюда получаем границы для t в зависимости от x.

1) Если x находится на отрезке от 0 до 1, то t меняется от 0 до x.  
Тогда плотность суммы равна интегралу от 0 до x от 2·e^(−2t) dt, что даёт значение 1 − e^(−2x).

2) Если x больше 1, то t меняется от x − 1 до x.  
Тогда плотность суммы равна интегралу от x − 1 до x от 2·e^(−2t) dt, что даёт значение e^(−2(x−1)) − e^(−2x).

3) Если x меньше 0, то сумма ξ + η не может принять такое значение, поэтому плотность равна 0.

---

### Ответ

Плотность распределения суммы Z = ξ + η равна:
- 0 при x < 0;
- 1 − e^(−2x) при 0 ≤ x ≤ 1;
- e^(−2(x−1)) − e^(−2x) при x > 1.
